```python
def add_sub(x, y):   
    c=x+y
    d=x-y
    return c,d  
r, r1 = add_sub(5, 6)
print(r)
print(r1)
print(type(r))
print(type(r1))
```

```python
def add_sub_mul_div(x, y):   
    c=x+y
    d=x-y
    e=x*y
    f=x//y
    return c,d,e,f  
r, r1, r2, r3 = add_sub_mul_div(5, 6)
print(r)
print(r1)
print(r2)
print(r3) 
```
