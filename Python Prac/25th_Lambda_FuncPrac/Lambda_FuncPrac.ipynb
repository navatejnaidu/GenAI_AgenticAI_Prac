```python
def count(lst): 
    
    even = 0
    odd = 0
    
    for i in lst:
        if i%2 == 0:
            even += 1
        else:
            odd +=1 
    return even,odd

lst = [10, 9, 8, 50, 34, 100] 
even, odd = count(lst)         

print(even)
print(odd)
```

```python
def div(a,b):
    print(a / b)

def div_decorator(func):
    def inner(a,b):
        if a<b:
            a,b = b,a
        return func(a,b)
    return inner

div = div_decorator(div)

div(2,4)     
```
